## 1. 1匹つくって、覚えさせて、聞いてみる
迷路で「右・右・下」と進むことを3回見せます。

```python
frog = Frog()
frog.learn([
    ["右", "右", "下"],
    ["右", "右", "下"],
    ["右", "右", "下"],
])
print(frog.predict(["右", "右"]))
```

## 2. 一番だけじゃなく、候補と確率を見る
`predict()` は1位だけ。`top()` と `probabilities()` で分布が見えます。

```python
print(frog.top(["右", "右"]))
print(frog.probabilities(["右"]))
```

`<END>` は「ここで終わりそう」という意味の特別な記号です。

```python
print(frog.predict(["右", "右", "下"]))
print(frog.predict(["右", "右", "下"]) is END)
```

## 3. 中身をのぞく: `show()`
点数が **直接のつながり・履歴・並び・痕跡** のどこから来たかを表にします。

```python
frog.show(["右", "右"])
```

## 4. 何を覚えたか: `edges()`

```python
print(frog.edges("右"))
```

## 5. 覚えさせ方いろいろ
- 文字列 → 1文字ずつ
- リスト → 1要素が1記号(「停止」は分かれない)
- リストのリスト → 何本もの系列

```python
f1 = Frog().learn("あいうえお")
print(f1.predict("あいう"))

f2 = Frog().learn(["正常", "正常", "警告", "停止"])
print(f2.predict(["正常", "警告"]))

f3 = Frog().learn([["おはよう", "ございます"], ["おやすみ", "なさい"]], epochs=5)
print(f3.predict(["おやすみ"]))
```

## 6. 続きをしゃべらせる: `generate()`
`greedy=True` なら毎回1位を選び、既定では確率どおりにくじを引きます(`seed` で再現できます)。

```python
print(frog.generate(["右"], greedy=True))
for s in range(3):
    print(frog.generate(["右"], seed=s))
```

## 7. 保存して、あとで呼び戻す

```python
frog.save("my_frog.zip")
again = Frog.load("my_frog.zip")
print(again.top(["右", "右"]) == frog.top(["右", "右"]))
```
